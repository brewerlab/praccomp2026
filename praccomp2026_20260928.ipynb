```python
def temp_cov(temp, scale):
    if scale.upper()=="C" or scale.upper()=='CELSIUS':
        print(str(temp)+" C = "+str((9/5)*float(temp)+32)+" F") 
    elif scale.upper()=='F' or scale.upper()=='FAHRENHEIT':
        print(str(temp)+ ' F = '+str((5/9)*(float(temp)-32))+' C')
    else:
        print("You made a mistake")
```

```python
transcriptome = "data/Emedia_Trinity.fasta.transdecoder.pep"
outdir = "data/Emedia_out/"

import os
os.makedirs(outdir, exist_ok=True)
```

```python
#Go to http://practicalcomputing.org/aminoacid.html, copy table, and paste in gedit
#Use regex to parse to 'single letter symbol':mol weight,
#.+(.)\t(\d+\.\d+)$ -> '\1':\2,
#Paste entries into dictionary
#! /usr/bin/env python

AminoDict={
	'A':89.09,
	'R':174.20,
	'N':132.12,
	'D':133.10,
	'C':121.15,
	'Q':146.15,
	'E':147.13,
	'G':75.07,
	'H':155.16,
	'I':131.17,
	'L':131.17,
	'K':146.19,
	'M':149.21,
	'F':165.19,
	'P':115.13,
	'S':105.09,
	'T':119.12,
	'W':204.23,
	'Y':181.19,
	'V':117.15,
	'X':0.0,
	'-':0.0,
	'*':0.0
}

#import sys
import os
#os.remove("temp1.txt")
fp1=open(transcriptome, 'r')
fp2=open(outdir+"temp1.txt", 'w+')

lines=fp1.readlines()

for line in lines:
	line=line.strip()

	if line.startswith(">") and line.find("complete")!=-1:
		segs=line.split(" ")
		fp2.write(segs[0]+"\tcomplete\t")
	elif line.startswith(">") and line.find("partial")!=-1:
		segs=line.split(" ")
		fp2.write(segs[0]+"\tpartial\t")
	elif line.startswith(">") and line.find("internal")!=-1:
		segs=line.split(" ")
		fp2.write(segs[0]+"\tinternal\t")
	else:
		fp2.write(line+"\n")
fp1.close()
fp2.close()

fp3=open(outdir+"temp1.txt", 'r')
fp4=open(outdir+"protein_OUTPUT.csv", 'w+')

lines=fp3.readlines()
lines.sort()

count_complete=0
count_partial=0
count_internal=0

for line in lines:
	line=line.replace(">", "")
	line=line.strip()
	if len(line)>1:
		segs=line.split("\t")
		if segs[1]=="complete":
			count_complete+=1
			AAseq=str(segs[2])
			MW=0
			for each in AAseq:
				MW+=AminoDict[each]
			#print segs[0], segs[1], MW
			fp4.write(segs[0]+"\t"+segs[1]+"\tMolecular_Weight:"+"\t"+str(MW)+"\t\n")
		elif segs[1]=="partial":
			count_partial+=1
		elif segs[1]=="internal":
			count_internal+=1
#print "complete\t",count_complete
#print "partial\t",count_partial
#print "internal\t",count_internal

fp3.close()
fp4.close()

fp5=open(outdir+"protein_OUTPUT.csv", "r")
fp6=open(outdir+"protein_summary.txt", 'w+')
lines=fp5.readlines()

totalMW=0
for line in lines:
	line=line.strip()
	if len(line)>1:
		segs=line.split("\t")
		totalMW+=float(segs[3])
		fp6.write(line+"\n")
avgMW=totalMW/count_complete
fp6.write("complete_sequence_number: "+str(count_complete)+"\n"+"partial_sequence_number: "+str(count_partial)+"\n"+"internal_sequence_number: "+str(count_internal)+"\n"+"complete_sequence_average_Molecular_Weight: "+str(avgMW)+"\n")
#print totalMW
#print avgMW

print(("complete_sequence_number: "+str(count_complete)+"\n"+"partial_sequence_number: "+str(count_partial)+"\n"+"internal_sequence_number: "+str(count_internal)+"\n"+"complete_sequence_average_Molecular_Weight: "+str(round(avgMW, 2))+"\n"))

fp5.close()
fp6.close()
os.remove(outdir+"temp1.txt")
os.remove(outdir+"protein_OUTPUT.csv")
```
